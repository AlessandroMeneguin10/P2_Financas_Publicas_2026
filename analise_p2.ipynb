# P2 Finanças do Setor Público — Suíça, Quixadá e Canindé
Este notebook inicia a auditoria da série da P1. Os dados transcritos são apenas referência e não devem ser tratados como oficiais até validação. Fontes oficiais a integrar: SICONFI/Tesouro Nacional, IBGE e FFA/EFV da Suíça.

## 1. Ambiente e caminhos

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
ROOT = Path.cwd()
DATA = ROOT / "dados"
OUT = ROOT / "resultados"
OUT.mkdir(exist_ok=True)
pd.set_option("display.max_columns", 100)
```

## 2. Carregar a série de referência da P1

```python
p = DATA / "base_p1_transcrita_nao_validada.csv"
df = pd.read_csv(p)
for col in ["receita_total_p1_valor_sem_validacao", "despesa_total_p1_valor_sem_validacao"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
df.head(10)
```

## 3. Auditoria e ausência de dados

```python
print("Dimensão:", df.shape)
print("Cobertura:", df.ano.min(), "-", df.ano.max())
print("Municípios:", df.municipio.unique())
print("Duplicatas ano-município:", df.duplicated(["ano", "municipio"]).sum())
display(df.isna().sum().to_frame("ausentes"))
```

## 4. Visualização descritiva da base anterior (não validada)

```python
for col, title in [("receita_total_p1_valor_sem_validacao", "Receita P1 — não validada"), ("despesa_total_p1_valor_sem_validacao", "Despesa P1 — não validada")]:
    fig, ax = plt.subplots(figsize=(10,5))
    for mun, g in df.groupby("municipio"):
        g = g.sort_values("ano")
        ax.plot(g.ano, g[col], marker="o", label=mun)
    ax.set(title=title, xlabel="Ano", ylabel="Valor transcrito da P1")
    ax.grid(alpha=.25); ax.legend(); plt.show()
```
